# Calculates the total needed simulation time

## Load libraries

```python
import os
import libs.tools as tools
```

## Setup

```python
folder = 'Statistics'
```

## Count files, times and events

```python
statistic_files = os.listdir(folder)

files = 0
runtime = 0
runcoretime = 0
events = 0
arrivals = 0

for statistic_file in statistic_files:
    if not statistic_file.endswith(".xml"):
        continue
    absolute_statistic_file = os.path.abspath(folder + os.sep + statistic_file)

    tree, root = tools.load_xml_file(absolute_statistic_file)
    el1 = tools.get_element(root, 'SimulationData', absolute_statistic_file)
    el1a = tools.get_element(el1, 'StatisticsRunTime', absolute_statistic_file)
    el1b = tools.get_element(el1, 'StatisticsEvents', absolute_statistic_file)
    el1c = tools.get_element(el1, 'StatisticsThreads', absolute_statistic_file)

    el2 = tools.get_element(root, 'StatisticsInterArrivalTimesClients', absolute_statistic_file)
    el2a = tools.get_element(el2, 'StationData', absolute_statistic_file)

    files += 1
    runtime += int(el1a.text)  # type: ignore
    runcoretime += (int(el1a.text) * int(el1c.text) / 2)  # type: ignore
    events += int(el1b.text)  # type: ignore
    arrivals += int(el2a.attrib['Count'])  # type: ignore
```

```python
print("Simulated models:", files)
print("Runtime (per node):", round(runtime / 1000 / 3600 / 24, 2), "days")
print("Runtime (per physical core):", round(runcoretime / 1000 / 3600 / 24, 2), "days")
ex = 13
print("Simulated events:", round(events / (10**ex), 3), f'* 10^{ex}')
print("Client arrivals:", round(arrivals / (10**ex), 3), f'* 10^{ex}')
```
