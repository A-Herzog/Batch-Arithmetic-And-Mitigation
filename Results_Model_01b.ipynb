## Load libraries

```python
import numpy as np

import libs.tools as tools
```

## Load and process data

```python
# Number of operators needed in bS=1 case
c_needed = 16

# Multirun results
data = tools.prepare_statistics(c_needed, lambda rec: rec["mode"] == "multirun")
tools.load_statistics(data)
```

```python
bS = [1, 2, 4, 8, 16]
EW = [[record["E[W]"] for record in data if record["b_max"] == b] for b in bS]
```

## Calculate interquartial range and total range

```python
for x, y in zip(bS, EW):
    IQR = np.quantile(y, 0.75) - np.quantile(y, 0.25)
    TR = max(y) - min(y)
    m = np.mean(y)
    print(f"bS={x}\tmean={m:.2f}\tIQR={IQR:.2f} ({IQR / m * 100:.2f}%)\tTR={TR:.2f} ({TR / m * 100:.2f}%)")
```
