# Models 4 and 5 - Service batch size range (diagrams for paper)

## Load libraries

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

import libs.tools as tools

plt.style.use('seaborn-v0_8')
```

## Load and process data

```python
# Number of operators needed in bS=1 case
c_needed = 16

# bI=1
data_bI1 = tools.prepare_statistics(c_needed, lambda rec: rec["mode"] == "range" and rec["bI"] == 1)
tools.load_statistics(data_bI1)

# bI=1..4
data_bI1234 = tools.prepare_statistics(c_needed, lambda rec: rec["mode"] == "range"  and isinstance(rec["bI"], str))
tools.load_statistics(data_bI1234)
```

### Calculate E[W] increasement relative to $b_S$=16..16 case

```python
for rec in data_bI1:
    base = tools.find_same_with_other_b(data_bI1, rec, 1, c_needed, c_needed, None)
    rec["E[W]rel"] = rec["E[W]"] / base["E[W]"]
```

### Calculate E[W] increasement relative to $b_S$=16..16 and $b_I=(1,1,1,1)$ case

```python
for rec in data_bI1234:
    base = tools.find_same_with_other_b(data_bI1234, rec, "1111", c_needed, c_needed, None)
    rec["E[W]rel"] = rec["E[W]"] / base["E[W]"]
```

## Plot results

### Setup plot

```python
bI_rates_levels = [
    [1, 1, 1, 1],
    [1, 2, 3, 4],
    [4, 3, 2, 1],
    [1, 2, 2, 1],
    [2, 1, 1, 2]
]

colors = ["blue", "red", "green", "orange", "purple"]

bS_colors=[
    '#F00',
    '#E01',
    '#D02',
    '#C03',
    '#B04',
    '#A05',
    '#906',
    '#807',
    '#708',
    '#609',
    '#50A',
    '#40B',
    '#30C',
    '#20D',
    '#10E',
    '#00F'
]
```

### Plot E[W] (relative) for $b_I=1$ (Model 4)

```python
fig, ax = plt.subplots(4, 3, figsize=(12, 12), sharex=True, sharey=True)

for i in range(4):
    if i == 0:
        S = "Exp"
        S_name = "Exponential"
        CVS = 1
    else:
        S = "Log"
        S_name = "Log-Normal"
        CVS = 0.5 * i
    for j in range(3):
        rho = round(0.60 + j * 0.15, 2)
        plot_sim_data = []
        for rec in [r for r in data_bI1 if r["S"] == S and r["model_CV[S]"] == CVS and r["model_rho"] == rho]:
            b = rec["b_min"]
            name = rec["name_short"]
            plot_sim_data.append({"b": b, "name": name, "E[W]": rec["E[W]rel"], "rho": rec["rho"]})

        plot_sim_data = sorted(plot_sim_data, key=lambda rec: rec["b"])

        x = [rec["b"] for rec in plot_sim_data]
        y1_sim = [rec["E[W]"] for rec in plot_sim_data]
        y2_sim = [rec["rho"] for rec in plot_sim_data]

        ax1 = ax[i][j]

        if i == 0 and j == 0:
            ax1.plot(x, y1_sim, marker='o', label='$\\mathbf{{E}}[W]$ relative to $b_S=16$')
        else:
            ax1.plot(x, y1_sim, marker='o')
        ax1.set_xticks(x)
        ax1.yaxis.set_major_formatter(FuncFormatter(lambda y, _: '{:.0%}'.format(y)))
        if j == 0:
            ax1.set_ylabel("$\\mathbf{{E}}[W]$ relative to $b_S$=16", color="darkblue")
            ax1.tick_params(axis='y', colors='darkblue')

        if i == 3:
            ax1.set_xlabel("$b_{S,\\min}$")

        ax1.set_title(f"S={S_name},\n$\\rho_{{\\text{{nom}}}}$={rho * 100:0.0f}% and CV[$S_{{\\text{{batch}}}}$]={CVS}")

        ax2 = ax1.twinx()
        if i == 0 and j == 0:
            ax2.plot(x, y2_sim, 'r:', marker='s', label='Effective operator utilization $\\rho_{\\text{eff}}$')
        else:
            ax2.plot(x, y2_sim, 'r:', marker='s')
        if j == 2:
            ax2.set_ylabel("Effective operator utilization $\\rho_{\\text{eff}}$", color="red")
            ax2.tick_params(axis='y', colors='red')
        else:
            ax2.set_yticks([])
        ax2.set_ylim([0.6, 1])
        ax2.yaxis.set_major_formatter(FuncFormatter(lambda y, _: '{:.0%}'.format(y)))

fig.suptitle("Model 4: $\\mathbf{E}[W]$ relative to $b_S$=16 and effective utilization $\\rho_{\\text{eff}}$", fontsize=16, fontweight="semibold", y=0.95)
fig.legend(loc='outside lower center', ncols=3, bbox_to_anchor=(0.5, 0.02), fontsize=10)
fig.subplots_adjust(wspace=0.05, hspace=0.4)

# fig.savefig("images/Model04-EWrel.png", format="png", bbox_inches='tight', pad_inches=0)

# fig.savefig("images/BatchArithmeticAndMitigation-figure-results-Model04-EWrel.pdf", format="pdf", bbox_inches='tight', pad_inches=0)
```

### Plot batch sizes frequencies for $b_I=1$ (Model 4)

```python
fig, ax = plt.subplots(4, 3, figsize=(12, 12), sharex=True, sharey=True)

for i in range(4):
    if i == 0:
        S = "Exp"
        S_name = "Exponential"
        CVS = 1
    else:
        S = "Log"
        S_name = "Log-Normal"
        CVS = 0.5 * i
    for j in range(3):
        rho = round(0.60 + j * 0.15, 2)
        plot_sim_data = []
        for rec in [r for r in data_bI1 if r["S"] == S and r["model_CV[S]"] == CVS and r["model_rho"] == rho]:
            b = rec["b_min"]
            name = rec["name_short"]
            plot_sim_data.append({"b": b, "name": name, "batch_sizes": rec["batch_sizes"], "rho": rec["rho"]})

        plot_sim_data = sorted(plot_sim_data, key=lambda rec: rec["b"])

        x = [rec["b"] for rec in plot_sim_data]

        ax1 = ax[i][j]
        bottom=[0] * len(x)
        for k in range(11, 16):
            above = [rec["batch_sizes"][k] for rec in plot_sim_data]
            if i==0 and j==0:
                ax1.bar(x, above, bottom=bottom, label=f"$b_S={k+1}$ used", color=bS_colors[16-3*(16-k)], edgecolor='black', linewidth=1)
            else:
                ax1.bar(x, above, bottom=bottom, color=bS_colors[16-3*(16-k)], edgecolor='black', linewidth=1)
            bottom = [bottom[m] + above[m] for m in range(len(x))]

        ax1.set_xticks(x)
        ax1.yaxis.set_major_formatter(FuncFormatter(lambda y, _: '{:.0%}'.format(y)))
        if j == 0:
            ax1.set_ylabel("Frequency for service batch sizes")

        if i == 3:
            ax1.set_xlabel("$b_{S,\\min}$")

        ax1.set_title(f"S={S_name},\n$\\rho_{{\\text{{nom}}}}$={rho * 100:0.0f}% and CV[$S_{{\\text{{batch}}}}$]={CVS}")

fig.suptitle("Model 4: Frequency for actual service batch sizes", fontsize=16, fontweight="semibold", y=0.95)
fig.legend(loc='outside lower center', ncols=5, bbox_to_anchor=(0.5, 0.02), fontsize=10)
fig.subplots_adjust(wspace=0.05, hspace=0.4)

# fig.savefig("images/Model04-batch_sizes.png", format="png", bbox_inches='tight', pad_inches=0)

# fig.savefig("images/BatchArithmeticAndMitigation-figure-results-Model04-batch_sizes.pdf", format="pdf", bbox_inches='tight', pad_inches=0)
```

### Plot E[W] (relative) for $b_I=1,2,3,4$ (Model 5)

```python
def plot_EWrel(ax, bI, color, label):
    for i in range(4):
        if i == 0:
            S = "Exp"
            S_name = "Exponential"
            CVS = 1.0
        else:
            S = "Log"
            S_name = "Log-Normal"
            CVS = 0.5 * i
        for j in range(3):
            rho = round(0.60 + j * 0.15, 2)
            plot_sim_data = []
            for rec in [r for r in data_bI1234 if r["S"] == S and r["model_CV[S]"] == CVS and r["model_rho"] == rho and r["bI"] == bI]:
                b = rec["b_min"]
                name = rec["name_short"]
                plot_sim_data.append({"b": b, "name": name, "E[W]": rec["E[W]rel"]})

            plot_sim_data = sorted(plot_sim_data, key=lambda rec: rec["b"])

            x = [rec["b"] for rec in plot_sim_data]
            y_sim = [rec["E[W]"] for rec in plot_sim_data]

            ax1 = ax[i][j]

            if i==0 and j==0:
                ax1.plot(x, y_sim, marker='o', color=color, label=label)
            else:
                ax1.plot(x, y_sim, marker='o', color=color)
            ax1.set_xticks(x)
            ax1.yaxis.set_major_formatter(FuncFormatter(lambda y, _: '{:.0%}'.format(y)))
            if j == 0:
                ax1.set_ylabel("$\\mathbf{{E}}[W]$ relative to\n$b_I=(1,1,1,1)$ and $b_S$=16")

            if i == 3:
                ax1.set_xlabel("$b_{S,\\min}$")

            ax1.set_title(f"S={S_name},\n$\\rho_{{\\text{{nom}}}}$={rho * 100:0.0f}% and CV[$S_{{\\text{{batch}}}}$]={CVS}")
```

```python
fig, ax = plt.subplots(4, 3, figsize=(12, 12), sharex=True, sharey=True)

for i, (bI_rate, color) in enumerate(zip(bI_rates_levels, colors)):  # type: ignore
    bI_str = "".join(map(str, bI_rate))
    plot_EWrel(ax, bI_str, color, f"$b_I$={tuple(bI_rate)}")

fig.legend(loc='outside lower center', ncols=5, bbox_to_anchor=(0.5, 0.02), fontsize=10)
fig.suptitle("Model 5: $\\mathbf{E}[W]$ relative to $b_I=(1,1,1,1)$ and $b_S$=16", fontsize=16, fontweight="semibold", y=0.95)
fig.subplots_adjust(wspace=0.05, hspace=0.4)

# fig.savefig("images/Model05-EWrel.png", format="png", bbox_inches='tight', pad_inches=0)

# fig.savefig("images/BatchArithmeticAndMitigation-figure-results-Model05-EWrel.pdf", format="pdf", bbox_inches='tight', pad_inches=0)
```

### Plot batch sizes frequencies for $b_I=1,2,3,4$ (Model 5)

```python
fig, ax = plt.subplots(4, 3, figsize=(12, 12), sharex=True, sharey=True)

for i in range(4):
    if i == 0:
        S = "Exp"
        S_name = "Exponential"
        CVS = 1
    else:
        S = "Log"
        S_name = "Log-Normal"
        CVS = 0.5 * i
    for j in range(3):
        rho = round(0.60 + j * 0.15, 2)
        plot_sim_data = {}
        for rec in [r for r in data_bI1234 if r["S"] == S and r["model_CV[S]"] == CVS and r["model_rho"] == rho]:
            b = rec["b_min"]
            batch_sizes = rec["batch_sizes"]
            if b in plot_sim_data:
                plot_sim_data[b] = [plot_sim_data[b][k] + batch_sizes[k] for k in range(len(batch_sizes))]
            else:
                plot_sim_data[b] = batch_sizes

        x = sorted(list(plot_sim_data.keys()))

        ax1 = ax[i][j]
        bottom = [0] * len(x)
        for k in range(11, 16):
            above = [plot_sim_data[x_val][k]/5 for x_val in x]
            if i==0 and j==0:
                ax1.bar(x, above, bottom=bottom, label=f"$b_S={k+1}$ used", color=bS_colors[16-3*(16-k)], edgecolor='black', linewidth=1)
            else:
                ax1.bar(x, above, bottom=bottom, color=bS_colors[16-3*(16-k)], edgecolor='black', linewidth=1)
            bottom = [bottom[m] + above[m] for m in range(len(x))]

        ax1.set_xticks(x)
        ax1.yaxis.set_major_formatter(FuncFormatter(lambda y, _: '{:.0%}'.format(y)))
        if j == 0:
            ax1.set_ylabel("Frequency for service batch sizes")

        if i == 3:
            ax1.set_xlabel("$b_{S,\\min}$")

        ax1.set_title(f"S={S_name},\n$\\rho_{{\\text{{nom}}}}$={rho * 100:0.0f}% and CV[$S_{{\\text{{batch}}}}$]={CVS}")

fig.legend(loc='outside lower center', ncols=5, bbox_to_anchor=(0.5, 0.02), fontsize=10)
fig.suptitle("Model 5: Frequency for actual service batch sizes", fontsize=16, fontweight="semibold", y=0.95)
fig.subplots_adjust(wspace=0.05, hspace=0.4)

# fig.savefig("images/Model05-batch_sizes.png", format="png", bbox_inches='tight', pad_inches=0)

# fig.savefig("images/BatchArithmeticAndMitigation-figure-results-Model05-batch_sizes.pdf", format="pdf", bbox_inches='tight', pad_inches=0)
```
