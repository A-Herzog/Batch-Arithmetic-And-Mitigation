## Total number of model files

```python
number_of_statistic_files = 11924
```

## Load libraries

```python
import os
import re
```

## Load data

```python
folder = 'Statistics'
```

```python
files = [file for file in os.listdir(folder) if file.endswith(".xml")]
```

## Number of completed simulation runs

```python
print(len(files), " (", round(len(files) / number_of_statistic_files * 100, 1), "%)", sep="")
```

## **Model 1** (should be 160)

```python
count = 0
files_remaining = []

for file in files:
    if re.search("Statistics\\-Batch\\-exp\\-rho[\\d]+\\-b[\\d]+\\.xml", file) is not None:
        count += 1
        continue
    if re.search("Statistics\\-Batch\\-log\\-rho[\\d]+\\-CV[\\d\\.]+\\-b[\\d]+\\.xml", file) is not None:
        count += 1
        continue
    files_remaining.append(file)

files = files_remaining

print(count)
assert(count == 160)
```
